```python
n=int(input())
lst = list(map(int, input().split()))[:n]
for i in range(n-1, -1, -1):
    print(lst[i])
```

```python
class Solution:
    def missingNumber(self, nums):
        n = len(nums)
        return n * (n + 1) // 2 - sum(nums)
```

```python
for i in range(n):
    for j in range(0, n - i - 1):
        if lst[j] > lst[j + 1]:
            lst[j], lst[j + 1] = lst[j + 1], lst[j]
print(lst)
```
