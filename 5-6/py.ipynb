```python
# Create a tuple
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple[0])
print(len(my_tuple))
print(my_tuple[1:3])
a, b, c, d, e = my_tuple
print(a, b, c)
```

```python
# tuples are immutable, so convert to a list first
temp_list = list(my_tuple)
popped_item = temp_list.pop()
print("popped item:", popped_item)
print("remaining items:", tuple(temp_list))
```

```python
# create a set from the tuple
my_set = set(my_tuple)
print(my_set)
print("Is 3 in set?", 3 in my_set)
```
