```python
n = int(input())
count = 0
if n==0:
    count=1
elif n<0:
    n=-n
while n!=0:
    count+=1
    n//=10
print(count)
```

```python
n = int(input())
res=0
while n!=0:
    res=res*10+n%10
    n//=10
print(res)
```
