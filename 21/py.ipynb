```python
''' Electric bill
units = 500
100-300 -> 5rs per unit
300-500 -> 10 rs per unit
>500 15 rs per unit
>500 apply extra changes -> 500 rs 
450 - 100 -> 350
200*5 -> 1000 -> 15
150*10 -> 1500'''
units = 500
bill = 0

if units <= 300:
    bill = (units - 100) * 5

elif units <= 500:
    bill = (200 * 5) + ((units - 300) * 10)

else:
    bill = (200 * 5) + (200 * 10) + ((units - 500) * 15) + 500

print("Electric Bill =", bill)
```
