```python
n = int(input())

for i in range(1, n + 1):
    # Print spaces
    for k in range(n - i):
        print(" ", end="")

    # Print stars and hollow spaces
    for j in range(2 * i - 1):
        if i == n or j == 0 or j == 2 * i - 2:
            print("*", end="")
        else:
            print(" ", end="")

    print()
```
